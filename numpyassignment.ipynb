```python
import numpy as np
import pandas as pd

print("=" * 50)
print("PART 1: NUMPY ARRAY OPERATIONS")
print("=" * 50)
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("1D Array (temperatures_w1):", temperatures_w1)
```

```python
print("\n--- 2. Inspection and Properties ---")
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
print("\n--- 3. Array Operations ---")
temperatures_w1_fah = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_w1_fah)
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])
print("\n2D Array (temperatures):\n", temperatures)
```

```python
print("\n--- 6. Inspect and Slice the 2D Array ---")
print("2D Shape:", temperatures.shape)
print("2D Data Type:", temperatures.dtype)
print("2D Total Number of Elements:", temperatures.size)
print("Temperatures for each week:\n", temperatures)
print("Weekends (last two days) for both weeks:\n", temperatures[:, -2:])
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80], 
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print("Original Marks Series:\n", marks)
```

```python
print("\n--- 2. Indexing and Slicing ---")
print("Mark of 1st rank student (integer index position 0):", marks.iloc[0])
print("Marks of top 3 ranks using loc:\n", marks.loc['Rank1':'Rank3'])
print("Mark of 3rd rank student using iloc:", marks.iloc[2])
```

```python
high_scorers = marks[marks > 90]
print("Ranks where marks > 90:\n", high_scorers)
```

```python
print("\n--- 3. Manipulating Series ---")
marks['Rank1'] = 100
print("Modified 1st rank mark to 100:\n", marks)

marks = marks.drop('Rank5')
print("After removing the last rank:\n", marks)
cgpa = marks / 10
print("Computed CGPA (Marks / 10):\n", cgpa)


print("\n" + "=" * 50)
print("PART 3: PANDAS DATAFRAME")
print("=" * 50)
```

```python
data = {'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 
                        'Clothing', 'Electronics', 'Furniture', 'Clothing', 
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 
               'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}
transactions = pd.DataFrame(data)
```

```python
print("\n--- 2. Data Exploration ---")
print("Head:")
print(transactions.head())
print("\nTail:")
print(transactions.tail())
print("\nShape:", transactions.shape)
print("Column Names:", list(transactions.columns))
print("Data Types:\n", transactions.dtypes)

print("\nOnly 'ProductCategory' and 'Amount' columns:")
print(transactions[['ProductCategory', 'Amount']])

print("\nLast 3 columns using iloc:")
print(transactions.iloc[:, -3:])

north_high = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print("\nRows where Region is 'North' and Amount > 200:")
print(north_high)

print("\nValue counts for 'ProductCategory':")
print(transactions['ProductCategory'].value_counts())

print("\nUnique values in 'Region':")
print(transactions['Region'].unique())

print("\nMean amount grouped by 'Region':")
print(transactions.groupby('Region')['Amount'].mean())

print("\n--- 3. Manipulating the DataFrame ---")
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("Updated Amount for TransactionID 102 to 165:")
print(transactions[transactions['TransactionID'] == 102])

transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAdded 'Discount' column:")
print(transactions.head())

transactions = transactions[transactions['TransactionID'] != 109]
print("\nAfter removing row with TransactionID 109:")
print(transactions)

transactions = transactions.drop(columns=['Discount'])
print("\nAfter deleting the 'Discount' column:")
print(transactions)
```
